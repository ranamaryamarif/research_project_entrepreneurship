```python
def whisker(col, per1, per3):
  q1, q3 = np.percentile(col, [per1, per3])
  iqr = q3 - q1

  lower_bound = q1 - (1.5 * iqr)
  upper_bound = q3 + (1.5 * iqr)
  return lower_bound, upper_bound

lw, uw = whisker(data["Starting_Salary"], 25, 75)

print('Lower Bound', lw)
print('Upper Bound', uw)
```

```python
def status(skewness):
  results = {}
  for col, skew in skewness.items():
    if skew > 1:
      category = "High Positive / Right Skew"

    elif skew < -1:
      category = "High Negative / Left Skew"

    elif -0.5 <= skew <= 0.5:
      category = "Approximately Normal"

    else:
      category = "Moderate Skew"

    results[col] = {"skewness": skew, "category": category}

  return pd.DataFrame(results).T
```

```python
# Plot Loss Curve
def plot_loss_curve(train_losses, valid_losses, train_accuracies, valid_accuracies):
  plt.figure(figsize=(10, 5))
  plt.plot(train_losses, label="Train Loss", marker='o')
  plt.plot(valid_losses, label="Validation Loss", marker='o')
  plt.xlabel("Iterations")
  plt.ylabel("Loss")
  plt.title("Loss Curve")
  plt.legend()
  plt.show()

  # Plot Accuracy Curve
  plt.figure(figsize=(10, 5))
  plt.plot(train_accuracies, label="Train Accuracy", marker='o')
  plt.plot(valid_accuracies, label="Validation Accuracy", marker='o')
  plt.xlabel("Iterations")
  plt.ylabel("Accuracy")
  plt.title("Accuracy Curve")
  plt.legend()
  plt.show()
```

```python
def confusion_matrix_plot(best_model, X_test, Y_test):
  y_pred_test = best_model.predict(X_test)

  from sklearn.metrics import classification_report, confusion_matrix

  cm = confusion_matrix(Y_test, y_pred_test)

  plt.figure(figsize=(8, 6))
  sns.heatmap(cm, annot=True, fmt = 'd', cmap = 'Blues')
  plt.xlabel('Predicted Labels')
  plt.ylabel('True Labels')
  plt.title('Confusion Matrix')
  plt.show()

  print(accuracy_score(Y_test, y_pred_test))
  print(classification_report(Y_test, y_pred_test))

  return y_pred_test
```

```python
def auc_curve(Y_test, y_pred_test):
  fpr, tpr, thresholds = roc_curve(Y_test, y_pred_test)
  auc = roc_auc_score(Y_test, y_pred_test)

  plt.figure(figsize=(8, 6))
  plt.plot(fpr, tpr, label=f'AUC = {auc:.2f}')
  plt.plot([0, 1], [0, 1], 'k--')
  plt.legend()
  plt.xlabel('False Positive Rate')
  plt.ylabel('True Positive Rate')
  plt.title('ROC Curve')
  plt.show()
```

Logistic Regression

```python
from sklearn.metrics import log_loss, accuracy_score

# Parameter grid
param_grid = {
    'penalty': ['l1', 'l2'],
    'C': [0.001, 0.01, 0.1, 1, 10],
    'solver': ['liblinear']
}

# Track best model
best_valid_accuracy = 0
best_params = None

cv_folds=5
scoring='f1'
n_jobs=-1
random_state=42

# Iterate over hyperparameters
for penalty in param_grid['penalty']:
    for C in param_grid['C']:
        for solver in param_grid['solver']:
            # Define model
            model = LogisticRegression(penalty=penalty, C=C, solver=solver, max_iter=2000)
            model.fit(X_train, Y_train)

            # Predictions
            Y_train_pred = model.predict_proba(X_train)[:, 1]
            Y_valid_pred = model.predict_proba(X_valid)[:, 1]

            # Compute log loss
            train_loss = log_loss(Y_train, Y_train_pred)
            valid_loss = log_loss(Y_valid, Y_valid_pred)

            # Compute accuracy
            train_acc = accuracy_score(Y_train, model.predict(X_train))
            valid_acc = accuracy_score(Y_valid, model.predict(X_valid))

            # Store losses and accuracies
            train_losses.append(train_loss)
            valid_losses.append(valid_loss)
            train_accuracies.append(train_acc)
            valid_accuracies.append(valid_acc)

            # Save best performing model
            if valid_acc > best_valid_accuracy:
                best_valid_accuracy = valid_acc
                best_params = (penalty, C, solver)
                best_model = LogisticRegression(penalty = penalty, C = C, solver = solver, max_iter=2000)

print(f"Best parameters: {best_params} with Validation Accuracy: {best_valid_accuracy}")
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, log_loss, classification_report, confusion_matrix
import numpy as np

# Expanded parameter grid
param_grid = {
    'criterion': ['gini', 'entropy', 'log_loss'],
    'max_depth': [None, 5, 10, 20, 50],
    'min_samples_split': [2, 5, 10, 20],
    'min_samples_leaf': [1, 2, 5, 10],
    'max_features': [None, 'sqrt', 'log2'],
    'max_leaf_nodes': [None, 10, 50, 100],
    'min_impurity_decrease': [0.0, 0.01, 0.1],
    'splitter': ['best', 'random'],
    'class_weight': [None, 'balanced'],
    'ccp_alpha': [0.0, 0.01, 0.1]
}

# Track best model
best_model = None
best_params = None
best_valid_accuracy = 0
lowest_valid_loss = float("inf")
smallest_loss_gap = float("inf")

# Lists to store performance metrics
train_losses, valid_losses = [], []
train_accuracies, valid_accuracies = [], []
trained_models, trained_params = [], []

# Iterate over hyperparameters
for criterion in param_grid['criterion']:
    for max_depth in param_grid['max_depth']:
        for min_samples_split in param_grid['min_samples_split']:
            for min_samples_leaf in param_grid['min_samples_leaf']:
                for max_features in param_grid['max_features']:
                    for max_leaf_nodes in param_grid['max_leaf_nodes']:
                        for min_impurity_decrease in param_grid['min_impurity_decrease']:
                            for splitter in param_grid['splitter']:
                                for class_weight in param_grid['class_weight']:
                                    for ccp_alpha in param_grid['ccp_alpha']:
                                        # Define Decision Tree Model
                                        model = DecisionTreeClassifier(
                                            criterion=criterion,
                                            max_depth=max_depth,
                                            min_samples_split=min_samples_split,
                                            min_samples_leaf=min_samples_leaf,
                                            max_features=max_features,
                                            max_leaf_nodes=max_leaf_nodes,
                                            min_impurity_decrease=min_impurity_decrease,
                                            splitter=splitter,
                                            class_weight=class_weight,
                                            ccp_alpha=ccp_alpha
                                        )

                                        # Train model ## isi k ander he data split hota ha
                                        model.fit(X_train, Y_train)

                                        # Predictions (probabilities for log loss)
                                        Y_train_pred = model.predict_proba(X_train)[:, 1]
                                        Y_valid_pred = model.predict_proba(X_valid)[:, 1]

                                        # Compute log loss
                                        train_loss = log_loss(Y_train, Y_train_pred)
                                        valid_loss = log_loss(Y_valid, Y_valid_pred)

                                        # Compute accuracy
                                        train_acc = accuracy_score(Y_train, model.predict(X_train))
                                        valid_acc = accuracy_score(Y_valid, model.predict(X_valid))

                                        # Store trained models & parameters
                                        trained_models.append(model)
                                        trained_params.append({
                                            'criterion': criterion,
                                            'max_depth': max_depth,
                                            'min_samples_split': min_samples_split,
                                            'min_samples_leaf': min_samples_leaf,
                                            'max_features': max_features,
                                            'max_leaf_nodes': max_leaf_nodes,
                                            'min_impurity_decrease': min_impurity_decrease,
                                            'splitter': splitter,
                                            'class_weight': class_weight,
                                            'ccp_alpha': ccp_alpha
                                        })

                                        # Store losses and accuracies
                                        train_losses.append(train_loss)
                                        valid_losses.append(valid_loss)
                                        train_accuracies.append(train_acc)
                                        valid_accuracies.append(valid_acc)

                                        # Compute loss gap (overfitting indicator)
                                        loss_gap = abs(train_loss - valid_loss)

                                        # Select the best model based on:
                                        # 1. Highest validation accuracy
                                        # 2. Lowest validation loss
                                        # 3. Smallest gap between train and validation loss
                                        if valid_acc > best_valid_accuracy and valid_loss < lowest_valid_loss and loss_gap < smallest_loss_gap:
                                            best_valid_accuracy = valid_acc
                                            lowest_valid_loss = valid_loss
                                            smallest_loss_gap = loss_gap
                                            best_model = model
                                            best_params = trained_params[-1]

# Print best parameters
print(f"Best Parameters: {best_params}")
print(f"Best Validation Accuracy: {best_valid_accuracy}")
print(f"Lowest Validation Loss: {lowest_valid_loss}")
print(f"Smallest Loss Gap: {smallest_loss_gap}")
```

```python
# Hyperparameter grid (expanded with main parameters)
param_grid = {
    'n_estimators': [50, 100],              # Number of trees in the forest
    'max_depth': [5, 10, None],             # Maximum depth of the trees
    'min_samples_split': [2, 5],            # Minimum samples required to split an internal node
    'min_samples_leaf': [1, 2],             # Minimum samples required at a leaf node
    'max_features': ['sqrt', 'log2', None], # Number of features to consider for best split
    'bootstrap': [True, False],             # Whether bootstrap samples are used
    'criterion': ['gini', 'entropy'],       # Function to measure quality of a split
    'class_weight': [None, 'balanced'],     # Weighting for imbalanced classes
    'random_state': [21]                    # Seed for reproducibility
}

# Tracking results
train_losses, valid_losses = [], []
train_accuracies, valid_accuracies = [], []
loss_gaps = []

# Best model selection
best_valid_acc = 0
smallest_loss_gap = float("inf")
best_model = None
best_params = None

# Manual grid search
for n in param_grid['n_estimators']:
    for depth in param_grid['max_depth']:
        for min_split in param_grid['min_samples_split']:
            for min_leaf in param_grid['min_samples_leaf']:
                for max_feat in param_grid['max_features']:
                    for bootstrap in param_grid['bootstrap']:
                        for criterion in param_grid['criterion']:
                            for class_weight in param_grid['class_weight']:
                                model = RandomForestClassifier(
                                    n_estimators=n,
                                    max_depth=depth,
                                    min_samples_split=min_split,
                                    min_samples_leaf=min_leaf,
                                    max_features=max_feat,
                                    bootstrap=bootstrap,
                                    criterion=criterion,
                                    class_weight=class_weight,
                                    random_state=21
                                )
                                model.fit(X_train, Y_train)

                                # Predictions & Probabilities
                                y_train_prob = model.predict_proba(X_train)[:, 1]
                                y_valid_prob = model.predict_proba(X_valid)[:, 1]

                                # Log loss
                                train_loss = log_loss(Y_train, y_train_prob)
                                valid_loss = log_loss(Y_valid, y_valid_prob)

                                # Accuracy
                                train_acc = accuracy_score(Y_train, model.predict(X_train))
                                valid_acc = accuracy_score(Y_valid, model.predict(X_valid))

                                # Compute loss gap (indicator of overfitting)
                                loss_gap = abs(train_loss - valid_loss)

                                # Store results
                                train_losses.append(train_loss)
                                valid_losses.append(valid_loss)
                                train_accuracies.append(train_acc)
                                valid_accuracies.append(valid_acc)
                                loss_gaps.append(loss_gap)

                                # Best model selection criteria:
                                # 1. Highest validation accuracy
                                # 2. Smallest loss gap (low overfitting)
                                if valid_acc > best_valid_acc and loss_gap < smallest_loss_gap:
                                    best_valid_acc = valid_acc
                                    smallest_loss_gap = loss_gap
                                    best_params = (n, depth, min_split, min_leaf, max_feat, bootstrap, criterion, class_weight)
                                    best_model = model

# Print best parameters
print(f"\n✅ Best Parameters Found: {best_params}")
print(f"🔹 Best Validation Accuracy: {round(best_valid_acc, 3)}")
print(f"⚖️ Smallest Loss Gap (Train vs Validation): {round(smallest_loss_gap, 5)}")
```

Adaboost

```python
# Hyperparameter grid (expanded with main parameters)
param_grid = {
    'n_estimators': [50, 100],              # Number of trees in the forest
    'learning_rate': [0.01, 0.1, 1]                  # Seed for reproducibility
}

# Tracking results
train_losses, valid_losses = [], []
train_accuracies, valid_accuracies = [], []
loss_gaps = []

# Best model selection
best_valid_acc = 0
smallest_loss_gap = float("inf")
best_model = None
best_params = None

# Manual grid search
for n in param_grid['n_estimators']:
    for rate in param_grid['learning_rate']:
      model = AdaBoostClassifier(
          n_estimators=n, learning_rate=rate, random_state=21
      )
      model.fit(X_train, Y_train)
      y_train_prob = model.predict_proba(X_train)[:, 1]
      y_valid_prob = model.predict_proba(X_valid)[:, 1]

      # Log loss
      train_loss = log_loss(Y_train, y_train_prob)
      valid_loss = log_loss(Y_valid, y_valid_prob)

      # Accuracy
      train_acc = accuracy_score(Y_train, model.predict(X_train))
      valid_acc = accuracy_score(Y_valid, model.predict(X_valid))

      # Compute loss gap (indicator of overfitting)
      loss_gap = abs(train_loss - valid_loss)

      # Store results
      train_losses.append(train_loss)
      valid_losses.append(valid_loss)
      train_accuracies.append(train_acc)
      valid_accuracies.append(valid_acc)
      loss_gaps.append(loss_gap)

      # Best model selection criteria:
      # 1. Highest validation accuracy
      # 2. Smallest loss gap (low overfitting)
      if valid_acc > best_valid_acc and loss_gap < smallest_loss_gap:
        best_valid_acc = valid_acc
        smallest_loss_gap = loss_gap
        best_params = (n, rate)
        best_model = model

# Print best parameters
print(f"\n✅ Best Parameters Found: {best_params}")
print(f"🔹 Best Validation Accuracy: {round(best_valid_acc, 3)}")
print(f"⚖️ Smallest Loss Gap (Train vs Validation): {round(smallest_loss_gap, 5)}")
```

```python
plot_loss_curve(train_losses, valid_losses, train_accuracies, valid_accuracies)
```
